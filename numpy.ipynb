Numpy Arrays - iterating

```python
arr = np.arange(1,21)

for x in arr:
    print(x)
```

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])

for x in arr:
  print(x)

for x in arr:
  for y in x:
    print(y)
```

```python
arr = np.arange(1,13)
arr = arr.reshape(2,3,2)
print(arr)

for x in arr:
    for y in x:
        for z in y:
            print(z)
```
